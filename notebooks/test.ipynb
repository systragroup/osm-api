```python
def overpass_decorator(query: str):
	header = """
	[out:json][timeout:180];
	(
	"""
	footer = """ 
	);
	out body;
	>;
	out skel qt;
	"""
	return header + query + footer


def get_overpass_query(bbox: tuple[float, float, float, float], key='highway', tag_list: list[str] = []) -> str:
	"""
	bbox : (ymin, xmin, ymax, xmax)
	"""
	if len(tag_list) == 0:
		return f'way["{key}"]{bbox};\n'
	else:
		tags = '|'.join([tag for tag in tag_list])
		return f'way["{key}"~"{tags}"]{bbox};\n'
```

```python
query = get_overpass_query((1, 1, 1, 1))
```

```python
overpass_decorator(query)
```
